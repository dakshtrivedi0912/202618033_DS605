## Task 2 - Vectorized Arithmetic and Linear Algebra

```python
M1 = np.array([
    [1, 2],
    [3, 4]
])

M2 = np.array([
    [5, 6],
    [7, 8]
])

print("Matrix 1:")
print(M1)

print("Matrix 2:")
print(M2)

addition = M1 + M2
print("Addition:")
print(addition)

element_multiplication = M1 * M2
print("Element wise multiplication:")
print(element_multiplication)

matrix_multiplication = M1 @ M2
print("Matrix Multiplication:")
print(matrix_multiplication)
```

```python
transpose = M1.T
print("Transpose:")
print(transpose)

determinant = np.linalg.det(M1)
print("Determinant:", determinant)
```

```python
if determinant != 0:
    inverse = np.linalg.inv(M1)
    print("Inverse:")
    print(inverse)
    print("Verification:")
    print(np.allclose(M1 @ inverse, np.eye(2)))

else:
    print("Matrix is not invertible.")
```

## Task 3 – Normal Distribution and Histogram

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)

chosen_mean = 50
chosen_std = 10

data = np.random.normal(loc=chosen_mean, scale=chosen_std, size=1000)

sample_mean = np.mean(data)
sample_std = np.std(data)

print("Chosen Mean:", chosen_mean)
print("Chosen Standard Deviation:", chosen_std)

print("\nSample Mean:", sample_mean)
print("Sample Standard Deviation:", sample_std)

print("\nComparison:")
print("Difference in Mean:", abs(chosen_mean - sample_mean))
print("Difference in Standard Deviation:", abs(chosen_std - sample_std))

plt.figure(figsize=(10, 6))

sns.histplot(data,
         bins=30,
         edgecolor="black",
         alpha=0.7,
         kde=True)

plt.axvline(sample_mean,
            linestyle="--",
            linewidth=2,
            label=f"Mean = {sample_mean:.2f}")

plt.title("Histogram of a Normal Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.grid(True)
plt.legend()
plt.show()
```

## Task 5 - Filtering and Querying

```python
import pandas as pd

df = pd.read_csv("train.csv")

male_over_50 = df[(df["Sex"] == "male") & (df["Age"] > 50)]

print("1. Male passengers older than 50:",len(male_over_50))

female_first_class = df.query("Sex == 'female' and Pclass == 1")

survival_percentage = (female_first_class["Survived"].mean() * 100)

print("\n2. Female first-class passengers:",len(female_first_class))

print("Percentage who survived:",round(survival_percentage, 2), "%")

median_fare = df["Fare"].median()

group_3 = df[(df["Age"] >= 20) & (df["Age"] <= 40) & (df["Fare"] > median_fare) & (df["Survived"] == 1)]

print("\n3. Age 20-40, Fare above the median, and survived:",len(group_3))

group_4 = df[(df["SibSp"] == 0) & (df["Parch"] == 0) & (df["Age"] < 30) & (df["Survived"] == 0)]

print("\n4. Travelling alone, age < 30, and did not survive:",len(group_4))


southampton_median = df[df["Embarked"] == "S"]["Fare"].median()

group_5 = df[(df["Embarked"] == "S") & (df["Pclass"].isin([2, 3])) & (df["Fare"] > southampton_median)]

print("\n5. Embarked = 'S', Pclass 2 or 3, and Fare above Southampton median:",len(group_5))
```

## Task 6 - groupby() and Aggregation

```python
import pandas as pd

df = pd.read_csv("train.csv")

print("1. Survival Rate by Sex")
survival_by_sex = df.groupby("Sex")["Survived"].mean() * 100
print(survival_by_sex)

print("\n2. Survival Rate by Pclass")
survival_by_pclass = df.groupby("Pclass")["Survived"].mean() * 100
print(survival_by_pclass)

print("\n3. Average Age and Fare by Pclass")
avg_age_fare = df.groupby("Pclass")[["Age", "Fare"]].mean()
print(avg_age_fare)

print("\n4. Passenger Count and Survival Rate by Sex and Pclass")
sex_pclass = df.groupby(["Sex", "Pclass"]).agg(Passenger_Count=("PassengerId", "count"),Survival_Rate=("Survived", "mean")
)
sex_pclass["Survival_Rate"] = sex_pclass["Survival_Rate"] * 100
print(sex_pclass)

print("\n5. Passenger Count, Average Fare, and Survival Rate by Embarked")
embarked_stats = df.groupby("Embarked").agg(Passenger_Count=("PassengerId", "count"), Average_Fare=("Fare", "mean"), 
                                                 Survival_Rate=("Survived", "mean"))
embarked_stats["Survival_Rate"] = (embarked_stats["Survival_Rate"] * 100)
print(embarked_stats)
```

## Task 7 - Missing Values and Fare Outliers

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

titanic = pd.read_csv("train.csv")

print("Missing Values in Each Column")
missing_count = titanic.isnull().sum()
missing_percentage = (titanic.isnull().sum() / len(titanic)) * 100
missing_table = pd.DataFrame({"Missing Count": missing_count,"Missing Percentage": missing_percentage})
print(missing_table)


plt.figure(figsize=(10, 6))
plt.bar(missing_count.index, missing_count.values)
plt.title("Missing Values in Titanic Dataset")
plt.xlabel("Columns")
plt.ylabel("Number of Missing Values")
plt.xticks(rotation=45)
plt.show()


print("\nMissing Age values before filling:")
print(titanic["Age"].isnull().sum())
titanic["Age"] = titanic["Age"].fillna(titanic["Age"].mean())

print("\nMissing Age values after filling:")
print(titanic["Age"].isnull().sum())

mean_df = pd.read_csv("train.csv")
median_df = pd.read_csv("train.csv")
mode_df = pd.read_csv("train.csv")
random_df = pd.read_csv("train.csv")

mean_df["Age"] = mean_df["Age"].fillna(mean_df["Age"].mean())
median_df["Age"] = median_df["Age"].fillna(median_df["Age"].median())
mode_df["Age"] = mode_df["Age"].fillna(mode_df["Age"].mode()[0])

missing_index = random_df[random_df["Age"].isnull()].index
random_values = np.random.choice(random_df["Age"].dropna(),size=len(missing_index))

random_df.loc[missing_index,"Age"] = random_values

print("\nMissing Age values after different imputations:")
print("Mean:", mean_df["Age"].isnull().sum())
print("Median:", median_df["Age"].isnull().sum())
print("Mode:", mode_df["Age"].isnull().sum())
print("Random:", random_df["Age"].isnull().sum())

Q1 = titanic["Fare"].quantile(0.25)
Q3 = titanic["Fare"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - (1.5 * IQR)
upper_bound = Q3 + (1.5 * IQR)
outliers = titanic[
    (titanic["Fare"] < lower_bound) |
    (titanic["Fare"] > upper_bound)
]

print("\nFare Outlier Analysis")
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of Outliers:", len(outliers))
```

## Task 8 - Features and Pivot Table

```python
import pandas as pd

titanic = pd.read_csv("train.csv")

titanic["FamilySize"] = (
    titanic["SibSp"] +
    titanic["Parch"] +
    1
)

titanic["IsAlone"] = (
    titanic["FamilySize"] == 1
).astype(int)

print("FamilySize and IsAlone")

print(
    titanic[
        [
            "SibSp",
            "Parch",
            "FamilySize",
            "IsAlone"
        ]
    ].head()
)

pivot_table = pd.pivot_table(
    titanic,
    values="Survived",
    index="Sex",
    columns="Pclass",
    aggfunc="mean"
)

print("\nPivot Table (Mean Survival Rate)")

print(pivot_table)

highest_group = pivot_table.stack().idxmax()
highest_value = pivot_table.stack().max()

lowest_group = pivot_table.stack().idxmin()
lowest_value = pivot_table.stack().min()

print("\nHighest Survival Group:")

print(
    "Sex:",
    highest_group[0],
    ", Pclass:",
    highest_group[1],
    ", Survival Rate:",
    round(highest_value * 100, 2),
    "%"
)

print("\nLowest Survival Group:")

print(
    "Sex:",
    lowest_group[0],
    ", Pclass:",
    lowest_group[1],
    ", Survival Rate:",
    round(lowest_value * 100, 2),
    "%"
)
```

## Task 9 - Visualizations and Observations

```python
import pandas as pd
import matplotlib.pyplot as plt

titanic = pd.read_csv("train.csv")

titanic["Age"] = titanic["Age"].fillna(titanic["Age"].mean())

numeric_columns = titanic.select_dtypes(include=["int64", "float64"])

correlation = numeric_columns.corr(numeric_only=True)

plt.figure(figsize=(8, 6))

plt.imshow(correlation, cmap="coolwarm")

plt.colorbar()

plt.xticks(
    range(len(correlation.columns)),
    correlation.columns,
    rotation=45
)

plt.yticks(
    range(len(correlation.columns)),
    correlation.columns
)

plt.title("Correlation Heatmap")

plt.tight_layout()

plt.show()

corr_pairs = correlation.unstack()

corr_pairs = corr_pairs[corr_pairs != 1]

strongest_positive = corr_pairs.idxmax()
strongest_negative = corr_pairs.idxmin()

print("Strongest Positive Relationship:")
print(
    strongest_positive,
    "=",
    round(corr_pairs.max(), 3)
)

print("\nStrongest Negative Relationship:")
print(
    strongest_negative,
    "=",
    round(corr_pairs.min(), 3)
)

survival_by_sex = (
    titanic.groupby("Sex")["Survived"]
    .mean() * 100
)

plt.figure(figsize=(6, 5))

plt.bar(
    survival_by_sex.index,
    survival_by_sex.values
)

plt.title("Survival Rate by Sex")

plt.xlabel("Sex")

plt.ylabel("Survival Rate (%)")

plt.show()

print("\nMale Survival Rate:",
      round(survival_by_sex["male"], 2), "%")

print("Female Survival Rate:",
      round(survival_by_sex["female"], 2), "%")

survived = titanic[titanic["Survived"] == 1]

not_survived = titanic[titanic["Survived"] == 0]

plt.figure(figsize=(8, 6))

plt.scatter(
    not_survived["Age"],
    not_survived["Fare"],
    label="Did Not Survive",
    alpha=0.6
)

plt.scatter(
    survived["Age"],
    survived["Fare"],
    label="Survived",
    alpha=0.6
)

plt.title("Age vs Fare by Survival Status")

plt.xlabel("Age")

plt.ylabel("Fare")

plt.legend()

plt.show()
```

## Observations

1) Female passengers had a much higher survival rate (≈74%) than male passengers (≈19%).
2) Passenger class had a strong relationship with survival. First-class passengers survived more often than second- and third-class passengers.
3) Fare and passenger class showed the strongest negative correlation because higher-class passengers generally paid higher fares.
4) Fare contained several outliers, indicating that some passengers paid exceptionally high ticket prices.
5) The scatter plot showed that passengers who paid higher fares were more likely to survive.
6) Most passengers who did not survive were concentrated in the lower-fare range.
7) Age showed a weaker relationship with survival than variables such as sex, fare, and passenger class.
